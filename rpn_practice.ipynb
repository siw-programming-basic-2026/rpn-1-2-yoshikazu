```python
# 問題9
# "3 4 +" を stack を使って計算してみましょう。

expression = "3 4 +"

# TODO:
# expression を split して tokens にしましょう。

tokens = expression.split()

stack = []

print("tokens:", tokens)
print("stack:", stack)

# TODO:
# まず 3 を stack に入れる
print("TODO: 3 を stack に入れる")
stack.append(int(tokens[0]))
print(stack)


# TODO:
# 次に 4 を stack に入れる
print("TODO: 4 を stack に入れる")
stack.append(int(tokens[1]))
print(stack)

# TODO:
# + が来たら、stack から2つ取り出す
print("TODO: stack から2つ取り出す")
if tokens[2]=="+":
    b=stack.pop()
    a=stack.pop()

# TODO:
# calculate() で計算する
print("TODO: calculate() で計算する")
result=calculate(a,b,tokens[2])

# TODO:
# 結果を stack に戻す
print("TODO: 結果を stack に戻す")
stack.append(result)

print("最後のstack:", stack)
print(stack)
```

```python
# 問題10
# for文を使って、tokens を1つずつ処理しましょう。
#
# 今回は + だけに対応します。

def calculate(a, b, operator):
    if operator == "+":
        return a + b

expressions = [
    "3 4 +",      # 予想:
    "10 20 +",    # 予想:
    "3 +",        # 予想:
    "5 a +",      # 予想:
    "3 4 5 +",    # 予想:
]

operators = {"+"}



# TODO:
# expression を split して tokens にしましょう。

for expression in expressions:

    tokens = expression.split()
    stack = []

    print("tokens:",tokens)
    print("stack:",stack)

    for token in tokens:
        print("token:", token)
        print("stack before:", stack)

        if token in operators:
            # TODO:
            # stack から2つ取り出しましょう。
            #
            # b = stack.pop()
            # a = stack.pop()
            if len(stack) < 2:
                print("値が足りません")
                break
            
            else:
                print("TODO: stackから2つ取り出す")
                b = stack.pop()
                a = stack.pop()

            # TODO:
            # calculate() で計算しましょう。
            #
            # result = calculate(a, b, token)

                print("TODO: calculate()で計算する")
                result=calculate(a,b,token)

            # TODO:
            # 計算結果を stack に戻しましょう。
            #
            # stack.append(result)

                print("TODO: 結果をstackに戻す")
                stack.append(result)

        else:
            # TODO:
            # token を整数に変換して stack に入れましょう。
            #
            # stack.append(int(token))
            if token.isdigit():
                print("TODO: 数字をstackに入れる")
                stack.append(int(token))
            else:
                print("不正な入力です。計算を止めます。")
                break
            

        print("stack after:", stack)
        print("-----")
    if len(stack)!=1:
        print("値が2つ以上あるので演算子が足りません。")
    else:    
        print("最後のstack:", stack)
        print()
```
