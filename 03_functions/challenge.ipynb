```python
def binary(num):
    binary=""
    while num>0:
        binary=str(num%2)+binary
        num=num//2
    return binary

for num in range(1,256):
    print(f"{num}:{binary(num)}")
```

```python
# 172.31.0.126 2進数表示をお願いします。
ipbin=[]
ipa=[172,31,0,126]
for ip in ipa:
    ipbin.append(binary(ip))

print(ipbin)
```

```python
def fizzbuzz(num):
    rule=(num % 3==0,num % 5 ==0)
    if rule==(True,True):
        return "FizzBuzz"
    elif rule==(True,False):
        return "Fizz"
    elif rule==(False,True):
        return "Buzz"
    else:
        return str(num)
    
for num in range(1,31):
    print(fizzbuzz(num))

```

```python
def fizzbuzz(num):
    rule = (num % 3 == 0, num % 5 == 0)

    rules = {
        (True, True): "FizzBuzz",
        (True, False): "Fizz",
        (False, True): "Buzz",
        (False, False): str(num)
    }

    return rules[rule]

for num in range(1,31):
    print(fizzbuzz(num))
```
