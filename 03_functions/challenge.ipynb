```python
#お題　このセル内でrpn_app.pyを再現してください。10行から20行程度をコピペすれば動きます。

from rpn_app import run_file_input,run_manual_input



def main():
    while True:
        print("入力方法を選んでください")
        print("1: 手入力")
        print("2: ファイル入力")
        print("3: 終了")

        choice = input("番号を入力してください: ").strip()

        if choice == "1":
            run_manual_input()
        elif choice == "2":
            run_file_input()
        elif choice == "3":
            print("終了します")
            break
        else:
            print("ERROR:1または2または3を入力してください")


if __name__ == "__main__":
    main()
```
